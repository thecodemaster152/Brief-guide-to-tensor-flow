Question 2: Why the Name Tensor flow?

    A tensor is just a multi-dimensional array of numbers.

| Data Type                         | Tensor Dimension |
| --------------------------------- | ---------------- |
| Single number                     | 0-D tensor       |
| List of numbers                   | 1-D tensor       |
| Matrix                            | 2-D tensor       |
| Image (height × width × channels) | 3-D tensor       |
| Video dataset                     | 4-D tensor       |


    A tensor flow through these operations hence the name TensorFlow

    Input Data → Neural Network Layers → Loss Calculation → Optimization → Predictions
